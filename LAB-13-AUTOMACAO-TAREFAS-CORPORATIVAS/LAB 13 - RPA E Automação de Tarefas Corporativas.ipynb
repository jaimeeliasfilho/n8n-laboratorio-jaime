```python
import pandas as pd

solicitacoes = {
    "id_solicitacao": [1001, 1002, 1003, 1004, 1005],
    "cliente": [
        "Ana Souza",
        "Carlos Lima",
        "Mariana Costa",
        "Pedro Santos",
        "Fernanda Alves"
    ],
    "tipo_solicitacao": [
        "ATUALIZAR EMAIL",
        "ATUALIZAR TELEFONE",
        "ATUALIZAR EMAIL",
        "ATUALIZAR CADASTRO",
        "ATUALIZAR TELEFONE"
    ],
    "status": [
        "PENDENTE",
        "PENDENTE",
        "CONCLUIDA",
        "PENDENTE",
        "PENDENTE"
    ]
}

df_solicitacoes = pd.DataFrame(solicitacoes)

display(df_solicitacoes)
```

```python
# Separar apenas as solicitações pendentes
df_pendentes = df_solicitacoes[
    df_solicitacoes["status"] == "PENDENTE"
]

print("Solicitações que serão processadas pelo robô:")
display(df_pendentes)

print("Total de tarefas pendentes:", len(df_pendentes))
```

```python
# Criar uma lista de tarefas para o robô
fila_execucao = df_pendentes.to_dict(orient="records")

# Conferir a primeira tarefa
print("Primeira tarefa da fila:")
print(fila_execucao[0])

print("\nTotal de tarefas na fila:", len(fila_execucao))
```

```python
# Simular o processamento individual das solicitações

for tarefa in fila_execucao:
    print("--------------------------------")
    print("ROBÔ INICIANDO PROCESSAMENTO")
    print("ID:", tarefa["id_solicitacao"])
    print("Cliente:", tarefa["cliente"])
    print("Tarefa:", tarefa["tipo_solicitacao"])
    print("Status atual:", tarefa["status"])
    print("Resultado: TAREFA PROCESSADA")
```

```python
from playwright.async_api import async_playwright

async def processar_fila_rpa():
    resultados = []

    async with async_playwright() as p:
        navegador = await p.chromium.launch(headless=True)
        pagina = await navegador.new_page()

        await pagina.set_content(html_formulario)

        for tarefa in fila_execucao:

            await pagina.locator("#id_solicitacao").fill(
                str(tarefa["id_solicitacao"])
            )

            await pagina.locator("#cliente").fill(
                tarefa["cliente"]
            )

            await pagina.locator("#tipo_solicitacao").fill(
                tarefa["tipo_solicitacao"]
            )

            await pagina.locator("button").click()

            confirmacao = await pagina.locator("#resultado").inner_text()

            resultados.append({
                "id_solicitacao": tarefa["id_solicitacao"],
                "cliente": tarefa["cliente"],
                "resultado": confirmacao
            })

            print(confirmacao)

        await navegador.close()

    return resultados

resultados_rpa = await processar_fila_rpa()
```

```python
import pandas as pd

df_resultados_rpa = pd.DataFrame(resultados_rpa)

df_resultados_rpa["status_execucao"] = "SUCESSO"

total_processadas = len(df_resultados_rpa)
total_sucesso = len(
    df_resultados_rpa[
        df_resultados_rpa["status_execucao"] == "SUCESSO"
    ]
)

print("RELATÓRIO DE EXECUÇÃO - LAB 13")
print("--------------------------------")
print("Total de solicitações:", total_processadas)
print("Processadas com sucesso:", total_sucesso)
print("Falhas:", total_processadas - total_sucesso)

display(df_resultados_rpa)
```
