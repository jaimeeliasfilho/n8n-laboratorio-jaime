```python
import math

# ==========================================
# LAB 18 - PLANEJAMENTO INTELIGENTE DE ESTOQUE
# ==========================================

# 1. Dados simulados dos produtos
produtos = [
    {"id": "P001", "nome": "Teclado USB", "estoque": 12, "vendas": 30, "prazo": 5},
    {"id": "P002", "nome": "Mouse sem fio", "estoque": 8, "vendas": 45, "prazo": 7},
    {"id": "P003", "nome": "Monitor 24 polegadas", "estoque": 20, "vendas": 12, "prazo": 10},
    {"id": "P004", "nome": "Headset USB", "estoque": 5, "vendas": 30, "prazo": 4},
    {"id": "P005", "nome": "Webcam HD", "estoque": 18, "vendas": 9, "prazo": 6}
]

# 2. Parâmetros da análise
dias_historico = 30
dias_seguranca = 3
dias_cobertura = 30

# 3. Análise individual dos produtos
for produto in produtos:

    consumo_diario = produto["vendas"] / dias_historico

    ponto_reposicao = consumo_diario * (
        produto["prazo"] + dias_seguranca
    )

    estoque_desejado = consumo_diario * dias_cobertura

    precisa_repor = produto["estoque"] <= ponto_reposicao

    # 4. Regra de negócio para recomendação
    if precisa_repor:
        quantidade_recomendada = max(
            0,
            math.ceil(estoque_desejado - produto["estoque"])
        )
    else:
        quantidade_recomendada = 0

    # 5. Apresentação dos resultados
    print("\n" + "=" * 45)
    print(f"Produto: {produto['nome']}")
    print(f"Consumo diário: {consumo_diario:.2f}")
    print(f"Ponto de reposição: {ponto_reposicao:.2f}")
    print(f"Estoque atual: {produto['estoque']}")
    print(f"Precisa repor: {'SIM' if precisa_repor else 'NÃO'}")
    print(f"Quantidade recomendada: {quantidade_recomendada}")

print("\nAnálise de estoque concluída!")
```

```python
import math
import json

# 1. Dados simulados
produtos = [
    {"id": "P001", "nome": "Teclado USB", "estoque": 12, "vendas": 30, "prazo": 5},
    {"id": "P002", "nome": "Mouse sem fio", "estoque": 8, "vendas": 45, "prazo": 7},
    {"id": "P003", "nome": "Monitor 24 polegadas", "estoque": 20, "vendas": 12, "prazo": 10},
    {"id": "P004", "nome": "Headset USB", "estoque": 5, "vendas": 30, "prazo": 4},
    {"id": "P005", "nome": "Webcam HD", "estoque": 18, "vendas": 9, "prazo": 6}
]

# 2. Parâmetros
dias_historico = 30
dias_seguranca = 3
dias_cobertura = 30

# 3. Lista para armazenar os resultados
resultados = []

# 4. Processamento
for produto in produtos:

    consumo_diario = produto["vendas"] / dias_historico

    ponto_reposicao = consumo_diario * (
        produto["prazo"] + dias_seguranca
    )

    estoque_desejado = consumo_diario * dias_cobertura

    precisa_repor = produto["estoque"] <= ponto_reposicao

    if precisa_repor:
        quantidade_recomendada = max(
            0,
            math.ceil(estoque_desejado - produto["estoque"])
        )
    else:
        quantidade_recomendada = 0

    # 5. Organizar cada produto como um registro
    resultado = {
        "produto_id": produto["id"],
        "produto": produto["nome"],
        "estoque_atual": produto["estoque"],
        "consumo_diario": round(consumo_diario, 2),
        "ponto_reposicao": round(ponto_reposicao, 2),
        "precisa_repor": precisa_repor,
        "quantidade_recomendada": quantidade_recomendada,
        "status": "REPOSICAO_NECESSARIA" if precisa_repor else "ESTOQUE_NORMAL"
    }

    resultados.append(resultado)

# 6. Exibir resultado estruturado em JSON
print(json.dumps(resultados, indent=4, ensure_ascii=False))
```
